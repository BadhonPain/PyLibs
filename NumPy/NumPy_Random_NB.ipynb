```python
# What is a random number ?
# => Random number does NOT mean a different number every time. 
# Random means something that can not be predicted logically.
```

```python
from numpy import random
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

**Generate Random Numbers**

```python
print(random.randint(100)) # random integer from 0-100
print(random.rand())       # random float numbers from 0-1
```

```python
# Generate random arrays
print(random.randint(100, size=7))
print(random.randint(100, size=(3,4)))
print(f"Float array: {random.rand(5)}")
```

```python
# choice(array, size)
print(random.choice([1, 2, 3, 4, 5], size=4))
print(random.choice([1, 2, 3, 4, 5], size=(2,1)))
```

**Random Data Distribution**

```python
# Random Distribution
print(random.choice([1, 2, 3, 4, 5], p=[0.1, 0.2, 0, 0.5, 0.2], size=100))
# Here p is probability density function
# 3 will never occur bcz it's probability is 0 and 4 will occur almost every time
```

```python
arr = np.array([11, 47, 14, 87])
x = random.choice(arr, p= [0.5, 0.3, 0.1, 0.1], size=(2,5))
print(x)
```

```python
random.shuffle(arr)  # changes to the original array
print(arr)
```

```python
newArr = random.permutation(arr)  # keeps original array unchanged
print(f"New Array: {newArr}")
print(f"Original Array: {arr}")
```

```python
# displot() -> distribution plot : to see how the data is distributed
sns.displot([0, 1, 2, 3, 4, 5, 6], kind='kde')
# without kind parameter, the visualization will occur with histogram
plt.show()
```

**Normal Distribution**

```python
# also known as Gaussian Distribution
# random.normal(loc,scale, size)
# loc - (Mean) where the peak of the bell exists
# scale - (Standard Deviation) how flat the graph distribution should be.
```

```python
print(random.normal(size=(2,3)))
```

```python
x = random.normal(loc=1, scale=2, size=(3, 5))
print(x)
```

```python
# Visualization
sns.displot(x, kind='kde')
plt.show()
```

```python
# This is also called Bell Curve
sns.displot(random.normal(size=100), kind='kde')
plt.show()
```

**Binomial Distribution**

```python
# Binomial Distribution is a Discrete Distribution.
# n = number of trials
# p = probability of occurence for each trial
# size = size of returned array
```

```python
# Given 10 trials for coin toss generate 10 data points:
x = random.binomial(n=10, p=0.5, size=10)
print(x)
```

```python
# Visualization
sns.displot(random.binomial(n=10, p=0.5, size=1000))
plt.show()
```

```python
# Normal vs Binomial Distribution
data = {
  "normal": random.normal(loc=50, scale=5, size=1000),
  "binomial": random.binomial(n=100, p=0.5, size=1000)
}
sns.displot(data, kind="kde")

plt.show()
```

**Uniform Distribution**

```python
# Used to describe probability where every event has equal chances of occuring.
# e.g. Generation of random numbers.

# It has three parameters:
# low - lower bound - default 0.0
# high - upper bound - default 1.0
# size - The shape of the returned array
```

```python
x = random.uniform(size=(2, 3))
print(x)
```

```python
# Visualization
sns.displot(random.uniform(size=1000), kind="kde")
plt.show()
```

**Logistic Distribution**

```python
# Logistic Distribution is used to describe growth.
# Used extensively in machine learning in logistic regression, neural networks etc.

# It has three parameters:
# loc - mean, where the peak is. Default 0.
# scale - standard deviation, the flatness of distribution. Default 1.
# size - The shape of the returned array.
```

```python
x = random.logistic(loc=1, scale=2, size=(2, 3))
print(x)
```

```python
# Visualization
sns.displot(random.logistic(size=1000), kind='kde')
plt.show()
```

```python
# Logistic vs Normal Distribution
data = {
  "normal": random.normal(scale=2, size=1000),
  "logistic": random.logistic(size=1000)
}
sns.displot(data, kind="kde")

plt.show()
```

```python
# Insert the missing part to generate a logistic distribution with mean at 1 and standard deviation is 2
x = random.logistic(loc=1, scale=2, size=(3,5))
print(x)
```

```python
sns.displot(random.logistic(size=1000), kind='kde')
plt.show()
```
