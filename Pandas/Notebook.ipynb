**Series**

```python
import pandas as pd
```

```python
data = [100, 102, 104]

series = pd.Series(data)  # Series constructor to make a series of data
print(series)
```

```python
data = ["A", "B", "C", "D"]
series = pd.Series(data)
print(series)
```

```python
data = [100.1, 102.2, 104.3]
series = pd.Series(data, index = ["Num-1:", "Num-2:", "Num-3:"])  # customized indexing (by default 0,1,2....)
print(series)
```

```python
series = pd.Series(data, index = ["a","b","c"])
print(f"{series.loc["a"]}, {series.loc["b"]}, {series.loc["c"]}") # loc -> location by level
# print(series["d"])  # Key Error
```

```python
# iloc -> integer location
print(series.iloc[0])
print(series.iloc[1])  # iloc only takes integer
print(series.iloc[2])
```

```python
# Dictionary as series
ct_marks = {
    "Alice" : 12,
    "Bob" : 16,
    "Badhon" : 18,
    "Sakib" : 20,
    "Messi" : 6
    }
_marks = pd.Series(ct_marks)
_marks
```

**Data Cleaning**

```python
# Drop irrelevent column(s)

# titanic_df.drop(columns="Cabin") # Cabin info is not so significant to predict
titanic_df.drop(columns=["Cabin", "Ticket"]) # multiple column dropping at a time
```

```python
# Handling missing data(s)
# Our age and embarking port name is significant to make prediction but it have null values
titanic_df["Age"].isnull().sum()
```

```python
# 177 passengers age value is missing, so we will replace it with median value
titanic_df["Age"].skew()
```

```python
# if value of skew lies between -0.5 ~ 0.5, then in case of replacing mean can be 
# used instead median but median always perfect
titanic_df["Age"].fillna(titanic_df["Age"].median())
# Study about inplace and it's behavior
```

```python
titanic_df["Age"].isnull().sum() # real object isn't updated :')
```

```python
titanic_df["Embarked"].isnull().sum()
```

```python
# We will fill this segment with it's mode
titanic_df["Embarked"].fillna(titanic_df["Embarked"].mode())
```

```python
titanic_df["Embarked"].isnull().sum()
```

```python
# Dropping Row(s)

# Dropping the rows who's age are NaN
titanic_df.dropna(subset=["Age"])
```

```python
titanic_df.fillna({"Cabin" : "None"})
```

```python
titanic_df.fillna({"Age" : titanic_df["Age"].median()})
```

```python
# Standardize Text
titanic_df["Name"] = titanic_df["Name"].str.upper()
display(titanic_df)
```

```python
# Fix data types
titanic_df["Survived"] = titanic_df["Survived"].astype(bool)
display(titanic_df)
```
