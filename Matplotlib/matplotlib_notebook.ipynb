**Pyplot & plot() Function**

```python
# pyplot is a submodule under matplotlib
```

```python
# Draw a line in a diagram from position (0,0) to position (4,27):
x_points = np.array([0, 4])
y_points = np.array([0, 27])
plt.plot(x_points, y_points)

plt.show()
```

```python
# without line
plt.plot(x_points, y_points, 'o')
plt.show()
```

```python
# Multiple points plotting
x_points = np.array([1, 2, 3, 4])
# If we don't pass x_points explicitly also it will be treated like this ~ Default x points
y_points = np.array([8, 2, 5, 0])
plt.plot(x_points, y_points)
plt.show()
```

**Matplotlib Markers**

```python
plt.plot(x_points, y_points, marker = 'o')  # Every point marked as circle
plt.show()
```

```python
plt.plot(x_points, y_points, marker = '.')  # Every point marked as point
plt.show()
```

```python
plt.plot(x_points, y_points, 'o--b')  # - -> solid line, : -> dotted line, -- -> dashed line
plt.show()
```

```python
# Marker size and markeredgecolor 
plt.plot(x_points, y_points, 'o:c', ms = 15, mec = 'r')
plt.show()
```

```python
# markerfacecolor 
plt.plot(x_points, y_points, 'o:c', ms = 15, mfc = 'r')  # we also can use hexadecimal color code
plt.show()
```

```python
plot_cusomization = dict(marker = 'o',
                         markersize = 20,  # expansion of ms -> increases readability
                         markerfacecolor = '#FF69B4',  # expansion of mfc
                         markeredgecolor = '#FF69B4',  # expansion of mec
                         color = '#46FF46'
                         )
plt.plot(x_points, y_points, **plot_cusomization)
plt.show()
```

```python
# We also can add separate attributes but plot_customization will work as default
plt.plot(x_points, y_points, linewidth = 8, linestyle = 'dashed', **plot_cusomization)
#                              lw= 8          ls= 'dashed  also valid
plt.show()
```

```python
# Multiple Lines
_x1 = np.array([1, 3, 9, 2])
_y1 = np.array([6, 2, 7, 11])
_x2 = np.array([0, 1, 2, 3])
_y2 = np.array([7, 2, 5, 16])

plt.plot(_x1, _y1, _x2, _y2)
plt.show()
```

```python
plt.plot(_y1, linewidth = 4, **plot_cusomization)
plt.plot(_x2, linestyle= 'dashed', **plot_cusomization)
```

**Title & Label**

```python
x = np.array([0, 1, 2, 3])
y = np.array([6, 2, 7, 11])
plt.plot(x, y)
plt.title("Point Plotting", fontsize= 20, 
                            family= 'Arial',
                            fontweight= 'bold',
                            color= 'r')
plt.xlabel("X Points ----->", fontweight= 'bold')
plt.ylabel("Y Points ----->", color= 'g',
                            fontweight= 'bold')
plt.show()
```

```python
plt.plot(x, y)
plt.xticks(x)
plt.show()
```

```python
# Use of fontdict()
x = np.array([80, 85, 90, 95, 100, 105, 110, 115, 120, 125])
y = np.array([240, 250, 260, 270, 280, 290, 300, 310, 320, 330])

font1 = {'family':'serif','color':'blue','size':20}
font2 = {'family':'serif','color':'darkred','size':15}

plt.title("Sports Watch Data", fontdict = font1, loc= 'left')
plt.xlabel("Average Pulse", fontdict = font2)
plt.ylabel("Calorie Burnage", fontdict = font2)

plt.plot(x, y)
plt.show()
```

**Grid**

```python
plt.plot(x,y)
plt.title("Plotting with grid", fontdict=font1)
plt.grid()
plt.show()
```

```python
# Grid in a specific axis
plt.plot(x, y)
plt.grid(axis= 'x')
plt.show()
```

```python
# Customizing grid
plt.title("Sports Watch Data", fontdict = font1)
plt.xlabel("Average Pulse", fontdict = font2)
plt.ylabel("Calorie Burnage", fontdict = font2)

plt.plot(x, y)
plt.grid(color= 'hotpink', ls='dotted', lw= 1.5)
plt.show()
```

**Sub-Plot**

```python
# subplot(rows, columns, index) function
x = np.array([0, 1, 2, 3])
y = np.array([3, 8, 1, 10])

plt.subplot(1, 2, 1)  # plot 1
plt.title("Plot-1")
plt.plot(x, y)

x = np.array([0, 1, 2, 3])
y = np.array([10, 20, 30, 40])

plt.subplot(1, 2, 2)  # plot 2
plt.title("Plot-2")
plt.plot(x, y)

plt.show()
```

```python
# Draw 2 plots on top of each other:
plt.subplot(2, 1, 1)
plt.plot(x, y)

plt.subplot(2, 1, 2)
plt.plot(x**2, y - 5)

plt.show()
```

```python
# 4 plots 
# plt.subplot(2, 2, 1)
plt.subplot(221)
plt.plot([1, 2, 3])

# plt.subplot(2, 2, 2)
plt.subplot(222)
plt.plot([3, 2, 1])

# plt.subplot(2, 2, 3)
plt.subplot(223)
plt.plot([1, 3, 2])

# plt.subplot(2, 2, 4)
plt.subplot(224) # shorthand notation
plt.plot([2, 1, 3])

plt.show()
```

**Bar, Histogram, Pie-charts**

```python
x = np.array(["MANGO", "LITCHI", "GREEN APPLE", "ORANGE"])
y = np.array([150, 480, 100, 300])

plt.xlabel("Name")
plt.ylabel("Price")
plt.title("Analysis of Fruits prices")

plt.bar(x, y, width= 0.5) # controlling width
plt.show()
```

```python
# Horizontal Bar
plt.xlabel("Price")
plt.ylabel("Name")
plt.title("Analysis of Fruits prices")
colors = ['yellow', 'red', 'green', 'orange']
plt.barh(x, y, color=colors, height=0.3)  # here we use height, instead width
plt.show()
```

```python
# Histogram - Used to show the distribution of numerical data
scores = [45, 52, 61, 67, 70, 72, 75, 78, 81, 85, 88, 91]

plt.hist(scores)
plt.show()
# x-axis shows score ranges, and the y-axis shows how many students are in each range
```

```python
# Specifying range with bins and random
data = np.random.normal(50, 10, 1000)

plt.hist(data, bins=20)
plt.show()
```

```python
# Pie chart : shows how a whole is divided into parts.
y = [35, 25, 25, 15]

plt.pie(y)
plt.show()
```

```python
# Adding labels
sizes = [35, 25, 25, 15]
labels = ["Study", "Sleep", "Gaming", "Exercise"]

plt.pie(sizes, labels=labels)
plt.show()
```

```python
# To show percentage
plt.pie(
    sizes,
    labels=labels,
    autopct="%1.1f%%"
)

plt.show()
```

```python
# Start the first wedge at 90 degrees
plt.pie(sizes, labels=labels, startangle=90)
plt.title("Daily Activites", fontweight= 'bold')
plt.show()
```

```python
# 'explode' parameter
my_explode = [0.1, 0, 0, 0]
plt.pie(y, labels=labels, explode=my_explode, shadow=True)
plt.show()
```

```python
# Customizing colors and adding label charts
colors = ['hotpink', 'black', 'cyan', 'orange']
plt.pie(y, colors= colors, labels=labels)
# plt.legend()
plt.legend(title='Activites')
plt.show()
```
